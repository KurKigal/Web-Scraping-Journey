## 📡 Örnek 1: En Basit HTTP İsteği

Web scraping'in ilk adımı: **Bir web sayfasına istek göndermek**

Bu örnekte öğrenecekleriniz:
- HTTP GET isteği nasıl gönderilir
- Status kodları ne anlama gelir
- HTML içeriği nasıl alınır

```python
# Test için güvenli bir site kullanıyoruz
url = "https://httpbin.org/html"

print(f"📡 İstek gönderiliyor: {url}")

# HTTP GET isteği gönder
response = requests.get(url)

# Yanıtı kontrol et
print(f"📊 Durum kodu: {response.status_code}")
print(f"📄 İçerik türü: {response.headers.get('content-type')}")
print(f"📏 İçerik uzunluğu: {len(response.text)} karakter")

# İçeriğin ilk kısmını göster
print("\n🔍 İçeriğin ilk 300 karakteri:")
print("-" * 50)
print(response.text[:300] + "...")
```

### 💡 Status Kodları Rehberi

- **200**: ✅ Başarılı
- **404**: ❌ Sayfa bulunamadı
- **403**: 🚫 Erişim yasak
- **429**: ⏰ Çok fazla istek (rate limit)
- **500**: 💥 Server hatası

## 🔍 Örnek 2: HTML Parse Etme

Ham HTML metnini işleyebilir hale getiriyoruz.

**BeautifulSoup** ile öğrenecekleriniz:
- HTML'i nasıl parse ederiz
- Tag'leri nasıl buluruz
- Text içeriklerini nasıl çıkarırız

```python
# Önceki response'u kullanıyoruz
if response.status_code == 200:
    # HTML'i BeautifulSoup ile parse et
    soup = BeautifulSoup(response.text, 'html.parser')
    
    print("🎯 HTML Parse Edildi!\n")
    
    # Sayfa başlığını bul
    title = soup.find('title')
    if title:
        print(f"📰 Sayfa başlığı: {title.text}")
    
    # İlk h1 tag'ini bul
    h1 = soup.find('h1')
    if h1:
        print(f"🏷️  İlk başlık: {h1.text}")
    
    # Tüm p tag'lerini bul
    paragraphs = soup.find_all('p')
    print(f"\n📝 Toplam paragraf sayısı: {len(paragraphs)}")
    
    for i, p in enumerate(paragraphs, 1):
        print(f"   📄 Paragraf {i}: {p.text[:50]}...")
```

### 🛠️ BeautifulSoup Temel Metodları

| Metod | Açıklama | Örnek |
|-------|----------|-------|
| `find()` | İlk eşleşen elementi bulur | `soup.find('h1')` |
| `find_all()` | Tüm eşleşen elementleri bulur | `soup.find_all('p')` |
| `select()` | CSS selector kullanır | `soup.select('.class-name')` |
| `.text` | Elementin text içeriğini alır | `element.text` |
| `.get()` | Elementin attribute'unu alır | `element.get('href')` |

## 🌍 Örnek 3: Gerçek Site Scraping

Şimdi gerçek bir web sitesinden veri çekelim!

**Quotes to Scrape** - Scraping öğrenmek için tasarlanmış güvenli site

Bu örnekte öğrenecekleriniz:
- Class'lara göre element bulma
- Multiple element'leri işleme
- Structured data oluşturma

```python
# Scraping öğrenmek için tasarlanmış güvenli site
url = "http://quotes.toscrape.com/"

# Etik scraping: User-Agent belirtme
headers = {
    'User-Agent': 'Mozilla/5.0 (Educational Bot; Learning Purpose)'
}

print(f"📡 İstek gönderiliyor: {url}")
response = requests.get(url, headers=headers)

print(f"📊 Status: {response.status_code}")

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Quotes'ları bul (her quote bir div.quote içinde)
    quotes = soup.find_all('div', class_='quote')
    print(f"💬 Bulunan alıntı sayısı: {len(quotes)}")
    
    print("\n" + "="*60)
    print("📋 İLK 3 ALINTI:")
    print("="*60)
```

```python
# İlk 3 alıntıyı detaylı inceleyelim
for i, quote in enumerate(quotes[:3], 1):
    print(f"\n📌 ALINTI {i}:")
    print("-" * 40)
    
    # Alıntı metnini al
    text = quote.find('span', class_='text')
    if text:
        print(f"💭 Metin: {text.text}")
    
    # Yazarı al
    author = quote.find('small', class_='author')
    if author:
        print(f"✍️  Yazar: {author.text}")
    
    # Tag'leri al
    tags = quote.find_all('a', class_='tag')
    tag_list = [tag.text for tag in tags]
    print(f"🏷️  Tag'ler: {', '.join(tag_list)}")
    
    print()
```

## ⚠️ Örnek 4: Hata Yönetimi

Gerçek dünyada her şey planlandığı gibi gitmez!

Bu örnekte öğrenecekleriniz:
- HTTP hata kodlarını yönetme
- Network hatalarını yakalama
- Timeout kullanımı
- Try-except blokları

```python
# Farklı durumları test edeceğiz
test_urls = [
    ("http://quotes.toscrape.com/", "✅ Çalışan site"),
    ("http://quotes.toscrape.com/notfound", "❌ 404 hatası"),
    ("http://nonexistent-site-12345.com/", "🔌 DNS hatası"),
]

print("🧪 HATA YÖNETİMİ TESTLERİ")
print("=" * 50)

for url, description in test_urls:
    print(f"\n🔍 Test: {description}")
    print(f"🌐 URL: {url}")
    
    try:
        # Timeout belirle - çok uzun beklemeyi önler
        response = requests.get(url, timeout=5)
        
        # Status code kontrolü
        if response.status_code == 200:
            print("✅ Başarılı! Site erişilebilir.")
            soup = BeautifulSoup(response.text, 'html.parser')
            title = soup.find('title')
            if title:
                print(f"   📰 Başlık: {title.text[:50]}")
        
        elif response.status_code == 404:
            print("⚠️  Sayfa bulunamadı (404)")
        
        elif response.status_code >= 500:
            print(f"❌ Server hatası: {response.status_code}")
        
        else:
            print(f"⚠️  Beklenmeyen durum kodu: {response.status_code}")
            
    except requests.exceptions.Timeout:
        print("⏰ Timeout hatası - Site çok yavaş yanıt veriyor")
        
    except requests.exceptions.ConnectionError:
        print("🔌 Bağlantı hatası - Site erişilemez veya mevcut değil")
        
    except requests.exceptions.RequestException as e:
        print(f"❌ Genel istek hatası: {e}")
    
    # Etik scraping: İstekler arası bekleme
    time.sleep(1)
```

### 🛡️ Hata Yönetimi En İyi Pratikleri

1. **Her zaman timeout kullanın** - Sonsuz beklemeyi önler
2. **Status kodlarını kontrol edin** - 200 dışındaki durumları yönetin
3. **Try-except blokları kullanın** - Network hatalarını yakalayın
4. **Retry mekanizması** - Geçici hatalar için tekrar deneyin
5. **Log tutun** - Hataları kaydedin ve analiz edin

## 🚀 Örnek 5: Gelişmiş Veri Çıkarma ve Temizleme

Artık profesyonel seviyede veri çıkarma teknikleri öğreniyoruz!

Bu örnekte öğrenecekleriniz:
- CSS selector kullanımı
- Attribute'lardan veri alma
- Text temizleme teknikleri
- Structured data oluşturma

```python
# Quotes sitesinden gelişmiş veri çıkarma
url = "http://quotes.toscrape.com/"
headers = {'User-Agent': 'Educational Bot'}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Structured data için liste
    quotes_data = []
    
    # CSS selector kullanarak quotes'ları bul
    quotes = soup.select('div.quote')
    
    print(f"🎯 Toplam {len(quotes)} alıntı bulundu")
    print("\n📊 VERİ ÇIKARMA VE TEMİZLEME:")
    print("=" * 50)
    
    for i, quote in enumerate(quotes, 1):
        # 1. Direct text extraction
        text_element = quote.select_one('span.text')
        quote_text = text_element.text.strip() if text_element else ""
        
        # 2. Author bilgisi
        author_element = quote.select_one('small.author')
        author = author_element.text.strip() if author_element else ""
        
        # 3. Multiple elements - tags
        tag_elements = quote.select('a.tag')
        tags = [tag.text.strip() for tag in tag_elements]
        
        # 4. Text cleaning - tırnak işaretlerini temizle
        clean_text = quote_text.replace('"', '').strip()
        
        # 5. Additional data processing
        word_count = len(clean_text.split())
        char_count = len(clean_text)
        
        # Structured data olarak kaydet
        quote_data = {
            'id': i,
            'text': clean_text,
            'author': author,
            'tags': tags,
            'word_count': word_count,
            'char_count': char_count,
            'tag_count': len(tags)
        }
        
        quotes_data.append(quote_data)
    
    print(f"✅ {len(quotes_data)} alıntı başarıyla işlendi!")
```

```python
# İlk 2 alıntının detaylı analizini gösterelim
print("🔍 DETAYLI VERİ ANALİZİ:")
print("=" * 50)

for quote in quotes_data[:2]:
    print(f"\n📌 Alıntı #{quote['id']}:")
    print(f"   💭 Metin: {quote['text'][:80]}...")
    print(f"   ✍️  Yazar: {quote['author']}")
    print(f"   📊 İstatistikler:")
    print(f"      • Kelime sayısı: {quote['word_count']}")
    print(f"      • Karakter sayısı: {quote['char_count']}")
    print(f"      • Tag sayısı: {quote['tag_count']}")
    print(f"   🏷️  Tag'ler: {', '.join(quote['tags'])}")
```

```python
# Veri analizi ve özetler
print("\n📈 VERİ ÖZETİ VE ANALİZİ:")
print("=" * 50)

# Temel istatistikler
total_quotes = len(quotes_data)
total_words = sum(quote['word_count'] for quote in quotes_data)
avg_words = total_words / total_quotes if total_quotes > 0 else 0

print(f"📊 Genel İstatistikler:")
print(f"   • Toplam alıntı sayısı: {total_quotes}")
print(f"   • Toplam kelime sayısı: {total_words}")
print(f"   • Ortalama kelime sayısı: {avg_words:.1f}")

# En uzun ve en kısa alıntılar
longest_quote = max(quotes_data, key=lambda x: x['word_count'])
shortest_quote = min(quotes_data, key=lambda x: x['word_count'])

print(f"\n🏆 Rekortmenler:")
print(f"   📏 En uzun alıntı: {longest_quote['word_count']} kelime - {longest_quote['author']}")
print(f"   📐 En kısa alıntı: {shortest_quote['word_count']} kelime - {shortest_quote['author']}")

# Tag analizi
all_tags = []
for quote in quotes_data:
    all_tags.extend(quote['tags'])

from collections import Counter
tag_counts = Counter(all_tags)
most_common_tags = tag_counts.most_common(5)

print(f"\n🏷️  En Popüler Tag'ler:")
for tag, count in most_common_tags:
    print(f"   • {tag}: {count} kez kullanıldı")
```
